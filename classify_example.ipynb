### 1. 퀄리티(CQ) 기반 통합 영상 분류 (화질 3구간 + 1초 키프레임 + 소형 영상 분포 자동 판단)
- **`_1_표준인코딩_CQ22`**: 과화질 영상($r > 1.2$, 720p H.264 기준 1800kbps 초과) 및 적정화질이면서 키프레임만 불량($> 2.2$초)인 영상 (원본 해상도·비율 유지 + `CQ 22` + `1초 키프레임` 인코딩)
- **`_2_저화질_키프레임교정`**: 저화질($r < 0.8$)이면서 키프레임이 불량($> 2.2$초)인 영상 (노이즈 증폭 방지 + `1초 키프레임` 인코딩)
- **루트 경로 잔류 (원본 보존)**: 키프레임이 정상인 적정/저화질 영상 및 절감 효과가 미미한 소형 영상

```python
from src import *
classifier = VideoClassifier()
classifier.add_exception_rule(lambda v: 'keep_origin' in v.filename)
# 필요 시 과화질 경계 배수 즉시 변경 가능 (기본값 1.2 -> 720p H.264 기준 1800kbps)
# classifier.set_over_bitrate_ratio(1.2)
classifier.pseudo_classify_mode() # 실제로 분류(폴더 이동)하려면 주석 처리 후 실행

classifier.classify()
classifier.print(sanitize_emoji=True, sort_key=lambda c: [
    col(c[t['예상 절감']], desc),
    col(c[t['b-rate 비율']], desc),
    col(WindowsNameOrder(c[t['이름']])),
])
```
